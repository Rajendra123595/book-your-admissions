```python

import json

def ask_admissions(question, course="", location=""):
    results = search_colleges(course=course, location=location)

    # search_colleges displays results, so filter directly for JSON output
    matches = df.copy()

    if course.strip():
        matches = matches[
            matches["course_name"].str.contains(
                course.strip(), case=False, na=False, regex=False
            )
        ]

    if location.strip():
        matches = matches[
            matches["location"].str.contains(
                location.strip(), case=False, na=False, regex=False
            )
        ]

    colleges = []
    sources = []

    for _, row in matches.iterrows():
        college = {
            "college_name": row["college_name"],
            "course_name": row["course_name"],
            "location": row["location"],
            "annual_fee_inr": (
                None if pd.isna(row["annual_fee_inr"])
                or str(row["annual_fee_inr"]).strip() == ""
                else row["annual_fee_inr"]
            ),
        }
        colleges.append(college)

        url = str(row["official_url"]).strip()
        if url and url.lower() != "nan" and url not in sources:
            sources.append(url)

    response = {
        "answer": (
            f"Found {len(colleges)} matching college-course records in the dataset."
            if colleges else
            "I don't have matching college information in my current dataset. Please check with the college."
        ),
        "colleges": colleges,
        "sources": sources,
        "found_in_data": bool(colleges),
    }

    print(json.dumps(response, indent=2, ensure_ascii=False))
    return response

# Test it
response = ask_admissions(
    question="Which colleges offer B.Tech CSE?",
    course="CSE"
)
```

```python

import pandas as pd

def filter_colleges(
    course="",
    location="",
    max_fee=None
):
    results = df.copy()

    # Clean text fields for reliable filtering
    for col in ["college_name", "course_name", "location"]:
        results[col] = results[col].fillna("").astype(str).str.strip()

    # Filter by course
    if course.strip():
        results = results[
            results["course_name"].str.contains(
                course.strip(), case=False, na=False, regex=False
            )
        ]

    # Filter by location
    if location.strip():
        results = results[
            results["location"].str.contains(
                location.strip(), case=False, na=False, regex=False
            )
        ]

    # Convert fees to numeric values
    results["annual_fee_inr"] = pd.to_numeric(
        results["annual_fee_inr"], errors="coerce"
    )

    # Apply budget only when fee data is available
    if max_fee is not None:
        results = results[
            results["annual_fee_inr"].notna()
            & (results["annual_fee_inr"] <= max_fee)
        ]

    if results.empty:
        print("No verified matching records found.")
        print("Check your course, location, and fee data.")
        return results

    display(results[
        ["college_name", "course_name", "location",
         "annual_fee_inr", "official_url", "checked_on"]
    ])

    return results

# Test 1: CSE colleges
filter_colleges(course="CSE")

# Test 2: Colleges in Ongole
filter_colleges(location="Ongole")

# Test 3: CSE colleges with a maximum annual fee of ₹100,000
filter_colleges(course="CSE", max_fee=100000)
```

```python

def search_brochures(question, top_k=3):
    words = {
        word.lower().strip(".,?!:;()")
        for word in question.split()
        if len(word) > 2
    }

    results = []

    for doc in documents:
        text_lower = doc["text"].lower()
        score = sum(
            1 for word in words if word in text_lower
        )

        if score > 0:
            results.append((score, doc))

    results.sort(key=lambda item: item[0], reverse=True)

    if not results:
        print("No relevant text found in the uploaded brochures.")
        return []

    for score, doc in results[:top_k]:
        print(f"\nSource: {doc['source']} | Page: {doc['page']}")
        print(f"Keyword matches: {score}")
        print(doc["text"][:1200])
        print("-" * 50)

    return [doc for score, doc in results[:top_k]]
```

```python

def answer_from_brochure(question, top_k=2):
    # Retrieve relevant brochure pages
    retrieved_pages = search_brochures(question, top_k=top_k)

    if not retrieved_pages:
        return {
            "answer": "I don't know based on the uploaded brochure.",
            "sources": [],
            "found_in_data": False
        }

    # Keep the answer grounded in retrieved evidence
    evidence = []
    sources = []

    for doc in retrieved_pages:
        evidence.append(
            f"Page {doc['page']}:\n{doc['text']}"
        )

        sources.append({
            "source": doc["source"],
            "page": doc["page"]
        })

    return {
        "answer": (
            "Relevant information found in the uploaded brochure. "
            "Review the evidence below before giving a final answer."
        ),
        "evidence": evidence,
        "sources": sources,
        "found_in_data": True
    }


result = answer_from_brochure(
    "What B.Tech courses are offered at PACE?"
)

print("ANSWER:")
print(result["answer"])

print("\nEVIDENCE:")
for item in result.get("evidence", []):
    print(item[:1500])

print("\nSOURCES:")
for source in result["sources"]:
    print(f"{source['source']} — Page {source['page']}")

print("\nFound in data:", result["found_in_data"])
```

```python

def admissions_assistant(question, course="", location="", max_fee=None):
    import pandas as pd

    results = df.copy()

    # Filter by course
    if course.strip():
        results = results[
            results["course_name"].str.contains(
                course.strip(), case=False, na=False
            )
        ]

    # Filter by location
    if location.strip():
        results = results[
            results["location"].str.contains(
                location.strip(), case=False, na=False
            )
        ]

    # Filter by maximum annual fee
    if max_fee is not None:
        fees = pd.to_numeric(results["annual_fee_inr"], errors="coerce")
        results = results[fees.notna() & (fees <= max_fee)]

    colleges = results[
        [
            "college_name",
            "course_name",
            "location",
            "annual_fee_inr",
            "official_url"
        ]
    ].to_dict(orient="records")

    # Search brochure evidence
    brochure_pages = search_brochures(question, top_k=2)

    sources = [
        {
            "source": page["source"],
            "page": page["page"]
        }
        for page in brochure_pages
    ]

    return {
        "question": question,
        "colleges": colleges,
        "brochure_evidence": [
            page["text"][:1000] for page in brochure_pages
        ],
        "sources": sources,
        "found_in_data": bool(colleges or brochure_pages)
    }


result = admissions_assistant(
    question="What B.Tech courses are offered at PACE?",
    course="B.Tech CSE",
    location="Andhra Pradesh"
)

print("Question:", result["question"])
print("Matching colleges:", len(result["colleges"]))
print("Sources:", result["sources"])
print("Found in data:", result["found_in_data"])
```

```python

import re

STOP_WORDS = {
    "what", "which", "where", "when", "does", "the",
    "are", "for", "from", "with", "and", "can", "how",
    "offered", "please", "tell", "about"
}

def search_brochures(question, top_k=3):
    words = {
        word.lower()
        for word in re.findall(r"[a-zA-Z0-9&]+", question)
        if len(word) > 2 and word.lower() not in STOP_WORDS
    }

    results = []

    for doc in documents:
        text_lower = doc["text"].lower()
        matches = [word for word in words if word in text_lower]
        score = len(matches)

        if score > 0:
            results.append((score, doc))

    results.sort(key=lambda item: item[0], reverse=True)

    if not results:
        print("No relevant text found in the uploaded brochures.")
        return []

    selected = results[:top_k]

    for score, doc in selected:
        print(f"\nSource: {doc['source']} | Page: {doc['page']}")
        print("Matched keywords:", score)
        print(doc["text"][:1000])
        print("-" * 50)

    return [doc for score, doc in selected]
```

```python

def ask_admissions(question, course="", location="", max_fee=None):
    import pandas as pd

    # 1. Filter college records
    results = df.copy()

    if course.strip():
        results = results[
            results["course_name"].str.contains(
                course.strip(), case=False, na=False
            )
        ]

    if location.strip():
        results = results[
            results["location"].str.contains(
                location.strip(), case=False, na=False
            )
        ]

    if max_fee is not None:
        fees = pd.to_numeric(
            results["annual_fee_inr"], errors="coerce"
        )
        results = results[fees.notna() & (fees <= max_fee)]

    # 2. Retrieve relevant brochure pages
    pages = search_brochures(question, top_k=2)

    sources = [
        {
            "document": page["source"],
            "page": page["page"]
        }
        for page in pages
    ]

    # 3. Build a structured response
    answer = (
        "I found relevant information in the uploaded sources. "
        "Review the evidence and citations before relying on it."
        if pages or not results.empty
        else "I don't know based on the available data."
    )

    return {
        "answer": answer,
        "colleges": results[
            [
                "college_name",
                "course_name",
                "location",
                "annual_fee_inr",
                "official_url"
            ]
        ].to_dict(orient="records"),
        "sources": sources,
        "evidence": [
            {
                "document": page["source"],
                "page": page["page"],
                "text": page["text"][:1000]
            }
            for page in pages
        ],
        "found_in_data": bool(pages or not results.empty)
    }


response = ask_admissions(
    question="What B.Tech courses are offered at PACE?",
    course="B.Tech CSE",
    location="Andhra Pradesh"
)

print("ANSWER:")
print(response["answer"])

print("\nCOLLEGES FOUND:", len(response["colleges"]))

print("\nSOURCES:")
for source in response["sources"]:
    print(f"{source['document']} — Page {source['page']}")

print("\nEVIDENCE PREVIEW:")
for item in response["evidence"]:
    print(f"\nPage {item['page']}:")
    print(item["text"][:500])
```

```python

def check_college_fee(college_name):
    matches = df[
        df["college_name"].str.contains(
            college_name, case=False, na=False
        )
    ]

    if matches.empty:
        return {
            "answer": "I don't know. This college is not in the dataset.",
            "sources": [],
            "found_in_data": False
        }

    fee_records = []

    for _, row in matches.iterrows():
        fee = row["annual_fee_inr"]

        if pd.notna(fee):
            fee_records.append({
                "college": row["college_name"],
                "course": row["course_name"],
                "annual_fee_inr": float(fee),
                "source": row["official_url"]
            })

    if not fee_records:
        return {
            "answer": (
                "I don't know the annual fee because it has not "
                "been verified in the current dataset."
            ),
            "sources": [
                url for url in matches["official_url"].dropna().unique()
            ],
            "found_in_data": True
        }

    return {
        "answer": "Verified fee information is available in the dataset.",
        "fee_records": fee_records,
        "sources": [item["source"] for item in fee_records],
        "found_in_data": True
    }


import pandas as pd

test = check_college_fee("PACE Institute of Technology and Sciences")
print(test)
```

```python

def check_brochure_answerability(question):
    q = question.lower()

    # Terms that should appear in evidence for each topic
    topic_rules = {
        "hostel fee": ["hostel", "fee"],
        "tuition fee": ["tuition", "fee"],
        "admission documents": ["documents", "admission"],
        "application deadline": ["last date", "deadline", "apply by"],
        "scholarship eligibility": ["scholarship", "eligibility"]
    }

    if "hostel" in q and "fee" in q:
        topic = "hostel fee"
    elif ("tuition" in q or "cse" in q) and "fee" in q:
        topic = "tuition fee"
    elif "document" in q:
        topic = "admission documents"
    elif "last date" in q or "deadline" in q:
        topic = "application deadline"
    elif "scholarship" in q:
        topic = "scholarship eligibility"
    else:
        return {
            "answer": "I don't know based on the current evidence.",
            "found_in_data": False,
            "sources": []
        }

    required_terms = topic_rules[topic]
    matching_pages = []

    for doc in documents:
        text = doc["text"].lower()

        if all(term in text for term in required_terms):
            matching_pages.append({
                "document": doc["source"],
                "page": doc["page"]
            })

    if not matching_pages:
        return {
            "answer": (
                "I don't know based on the uploaded brochure. "
                "Please verify this information from the official "
                "college admissions source."
            ),
            "found_in_data": False,
            "sources": []
        }

    return {
        "answer": (
            "Potentially relevant text was found. "
            "Verify that it actually answers the question."
        ),
        "found_in_data": True,
        "sources": matching_pages
    }


for question in test_questions:
    result = check_brochure_answerability(question)
    print("\nQuestion:", question)
    print("Answer:", result["answer"])
    print("Found in data:", result["found_in_data"])
    print("Sources:", result["sources"])
    print("-" * 40)
```

```python

import gradio as gr
import pandas as pd

def search_interface(course, location, max_fee):
    results = df.copy()

    if course and course != "All courses":
        results = results[
            results["course_name"].str.contains(
                course, case=False, na=False
            )
        ]

    if location.strip():
        results = results[
            results["location"].str.contains(
                location.strip(), case=False, na=False
            )
        ]

    # Apply a budget only when the user enters one.
    if max_fee is not None:
        fees = pd.to_numeric(
            results["annual_fee_inr"], errors="coerce"
        )
        results = results[fees.notna() & (fees <= max_fee)]

    if results.empty:
        return "No matching records found. Unknown fees are not treated as zero."

    display_df = results[
        [
            "college_name",
            "course_name",
            "location",
            "annual_fee_inr",
            "official_url"
        ]
    ].copy()

    display_df["annual_fee_inr"] = display_df[
        "annual_fee_inr"
    ].apply(lambda x: "Not verified" if pd.isna(x) else x)

    return display_df


course_options = ["All courses"] + sorted(
    df["course_name"].dropna().unique().tolist()
)

demo = gr.Interface(
    fn=search_interface,
    inputs=[
        gr.Dropdown(
            choices=course_options,
            value="All courses",
            label="Course"
        ),
        gr.Textbox(
            label="Location",
            placeholder="e.g. Andhra Pradesh"
        ),
        gr.Number(
            label="Maximum annual fee (₹)",
            minimum=0,
            value=None
        )
    ],
    outputs=gr.Dataframe(label="Matching colleges"),
    title="Book Your Admissions — College Search",
    description=(
        "Search your current college dataset. "
        "Unverified fees are clearly identified."
    )
)

demo.launch(share=True)
```

```python
def search_colleges_api(course, location, eligibility):
    response = requests.post(
        public_url + "/search_colleges",
        json={
            "course": course,
            "location": location,
            "eligibility": eligibility
        },
        headers={"ngrok-skip-browser-warning": "true"},
        timeout=30
    )

    response.raise_for_status()
    return response.json()
```

```python
def search_ui(course, location, eligibility):
    try:
        data = search_colleges_api(course, location, eligibility)

        if not data["found_in_data"]:
            return [["No matching colleges found.", "", "", "", "", ""]]

        output = []

        for c in data["colleges"]:
            output.append([
                c["college_name"],
                c["course_name"],
                c["location"],
                c.get("eligibility") or "Not available",
                c.get("annual_fee_inr") or "Not available",
                c["official_url"]
            ])

        return output

    except requests.exceptions.RequestException as e:
        return [[f"API error: {e}", "", "", "", "", ""]]
```

```python
import requests
import pandas as pd

test_cases = [
    {
        "test": "All Andhra Pradesh CSE colleges",
        "course": "B.Tech CSE",
        "location": "Andhra Pradesh",
        "eligibility": ""
    },
    {
        "test": "JEE Main eligibility filter",
        "course": "B.Tech CSE",
        "location": "Andhra Pradesh",
        "eligibility": "JEE Main"
    },
    {
        "test": "No matching location",
        "course": "B.Tech CSE",
        "location": "Mumbai",
        "eligibility": ""
    },
    {
        "test": "No matching course",
        "course": "B.Tech Aerospace",
        "location": "Andhra Pradesh",
        "eligibility": ""
    }
]

test_results = []

for test in test_cases:
    try:
        response = requests.post(
            public_url + "/search_colleges",
            json={
                "course": test["course"],
                "location": test["location"],
                "eligibility": test["eligibility"]
            },
            headers={"ngrok-skip-browser-warning": "true"},
            timeout=30
        )

        data = response.json()

        test_results.append({
            "Test": test["test"],
            "HTTP Status": response.status_code,
            "Results": data.get("count", 0),
            "Status": (
                "PASS" if response.status_code == 200
                and isinstance(data.get("colleges"), list)
                else "FAIL"
            )
        })

    except Exception as e:
        test_results.append({
            "Test": test["test"],
            "HTTP Status": "Error",
            "Results": 0,
            "Status": str(e)
        })

test_report = pd.DataFrame(test_results)
display(test_report)
```
